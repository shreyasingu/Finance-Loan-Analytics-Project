```python
numeric_columns = df_clean.select_dtypes(include='number').columns

print("Numeric columns:")
print(list(numeric_columns))
```

**Find Outliers Using IQR**

```python
outlier_summary = []

for col in numeric_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outlier_count = ((df_clean[col] < lower_limit) |
                     (df_clean[col] > upper_limit)).sum()

    outlier_summary.append({
        'Column': col,
        'Outlier_Count': outlier_count
    })

outlier_df = pd.DataFrame(outlier_summary)

outlier_df = outlier_df.sort_values(
    'Outlier_Count',
    ascending=False
)

display(outlier_df)
```

**4A)State-wise Loan Status Analysis**

```python
print(df_clean.columns.tolist())
```

```python
print('addr_state' in df_clean.columns)
print('loan_status' in df_clean.columns)
```

```python
state_status = df_clean.groupby(['addr_state','loan_status']).size().reset_index(name='Total_Loans')

print(state_status.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

top_states = df_clean['addr_state'].value_counts().head(10).index

state_chart = df_clean[df_clean['addr_state'].isin(top_states)]

pivot_table = pd.crosstab(
    state_chart['addr_state'],
    state_chart['loan_status']
)

pivot_table.plot(
    kind='bar',
    figsize=(12,6)
)

plt.title('Top 10 States Loan Status Analysis')
plt.xlabel('State')
plt.ylabel('Total Loans')
plt.xticks(rotation=45)
plt.legend(title='Loan Status')
plt.show()
```

**4B) Month-wise Loan Status Analysis**

```python
df_clean['issue_month'] = df_clean['issue_d'].dt.month_name()

month_status = pd.crosstab(
    df_clean['issue_month'],
    df_clean['loan_status']
)

month_order = [
    'January','February','March','April','May','June',
    'July','August','September','October','November','December'
]

month_status = month_status.reindex(month_order)

month_status.plot(
    kind='bar',
    figsize=(12,6)
)

plt.title('Month-wise Loan Status Analysis')
plt.xlabel('Month')
plt.ylabel('Total Loans')
plt.xticks(rotation=45)
plt.legend(title='Loan Status')
plt.show()
```

Insight:

Loan performance varies across states.

Some states have higher numbers of fully paid loans, while others show more charged-off loans.

Monthly trends reveal seasonal patterns in loan issuance and repayment outcomes.

**5)Home Ownership vs Last Payment Date Stats**

```python
df_clean['last_pymnt_d'] = pd.to_datetime(df_clean['last_pymnt_d'], errors='coerce')
```

```python
df_clean['last_payment_year'] = df_clean['last_pymnt_d'].dt.year
```

```python
home_payment_date = df_clean.groupby(
    ['home_ownership','last_payment_year']
).size().reset_index(name='Total_Loans')
print(home_payment_date.head())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))

for ownership in home_payment_date['home_ownership'].unique():
    temp = home_payment_date[
        home_payment_date['home_ownership'] == ownership
    ]
    plt.plot(
        temp['last_payment_year'],
        temp['Total_Loans'],
        marker='o',
        label=ownership
    )

plt.title('Home Ownership vs Last Payment Date Stats')
plt.xlabel('Last Payment Year')
plt.ylabel('Total Loans')
plt.legend(title='Home Ownership')
plt.grid(True)
plt.show()
```

Insight:

Different home ownership categories (Mortgage, Rent, Own) show different payment patterns.

Mortgage and Rent categories often contain the largest number of borrowers.

Last payment trends help understand repayment activity across customer groups.
